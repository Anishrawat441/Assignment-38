# Task 3: Code Assistant Features
## Create functions for:
## 1. Code generation
## 2. Code expalanation
## 3. Bug fixing
## 4. Code optimization

```python

from langchain_ollama import OllamaLLM

# Initialize CodeLlama
llm = OllamaLLM(model="codellama")


# 1. Code Generation
def generate_code(requirement):
    prompt = f"""
    You are an expert Python developer.
    Write clean and working Python code
    for the following requirement:

    {requirement}
    """
    return llm.invoke(prompt)


# 2. Code Explanation
def explain_code(code):
    prompt = f"""
    Explain the following code line by line.
    Describe its logic, and output.

    Code:
    {code}
    """
    return llm.invoke(prompt)


# 3. Bug Fixing
def fix_bugs(code, error_message=""):
    prompt = f"""
    You are a Python debugging assistant.
    Find and fix errors in the following code.

    Code:
    {code}

    Error message:
    {error_message}

    Provide the corrected code and explain
    what was wrong.
    """
    return llm.invoke(prompt)


# 4. Code Optimization
def optimize_code(code):
    prompt = f"""
    Optimize the following Python code for
    readability, efficiency, and maintainability.

    Code:
    {code}

    Return the optimized code and explain
    the improvements.
    """
    return llm.invoke(prompt)

# Test all functions
print("\n1. CODE GENERATION")
print(generate_code(
    "Write a Python function to check prime numbers."
))

sample_code = """
def square(n):
    return n * n

print(square(5))
"""

print("\n 2. CODE EXPLANATION:")
print(explain_code(sample_code))

buggy_code = """
def add(a, b):
    return a - b

print(add(5, 3))
"""

print("\n 3. BUG FIXING")
print(fix_bugs(
    buggy_code,
    "The function should add numbers but returns their difference."
))

slow_code = """
result = []
for i in range(100):
    result.append(i * 2)
print(result)
"""

print("\n4. CODE OPTIMIZATION ")
print(optimize_code(slow_code))

```

```python
from langchain_core.prompts import PromptTemplate

# 1. Code Generation Template
generate_template = PromptTemplate(
    input_variables=["user_input"],
    template="""
You are an expert Python developer.

TASK:
Generate Python code for the following requirement.

REQUIREMENT:
{user_input}

OUTPUT FORMAT:
1. Approach
2. Python Code
3. Brief Explanation

Write clean, readable code.
Avoid unnecessary explanations.
"""
)

# 2. Code Explanation Template
explain_template = PromptTemplate(
    input_variables=["user_input"],
    template="""
You are a Python programming teacher.

TASK:
Explain the following code clearly.

CODE:
{user_input}

OUTPUT FORMAT:
1. Purpose of the Code
2. Line-by-Line Explanation
3. Key Concepts

Use simple language and readable formatting.
"""
)

# 3. Debugging Template
debug_template = PromptTemplate(
    input_variables=["user_input"],
    template="""
You are an expert Python debugging assistant.

TASK:
Identify and fix bugs in the following code.

CODE:
{user_input}

OUTPUT FORMAT:
1. Identified Errors
2. Reason for  Error
3. Corrected Code

If no bug and error say your code is completly fine.
"""
)

# 4. Code Optimization Template
optimize_template = PromptTemplate(
    input_variables=["user_input"],
    template="""
You are an expert Python code optimization assistant.

TASK:
Improve the following code.

CODE:
{user_input}

OUTPUT FORMAT:
1. Current Code Issues
2. Optimized Code
3. Improvements Made

Preserve the original behavior.
Prefer readable, maintainable code.
"""
)
```
